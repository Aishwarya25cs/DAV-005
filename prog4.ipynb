```python
# ============================================
# DATA PROCESSING
# 1. Handling Missing Values
# 2. Encoding Categorical Variables
# ============================================

import pandas as pd
from sklearn.preprocessing import LabelEncoder, OrdinalEncoder

# Step 1: Create a sample dataset
df = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva'],
    'Age': [20, 25, None, 30, 35],
    'Salary': [20000, 30000, 40000, None, 50000],
    'Experience': [1, None, 3, 4, None],
    'City': ['Delhi', 'Mumbai', 'Delhi', 'Chennai', 'Mumbai'],
    'Education': ['School', 'Bachelor', 'Master',
                  'Bachelor', 'School'],
    'Purchased': ['No', 'Yes', 'No', 'Yes', 'No']
})

print("ORIGINAL DATASET:")
print(df)

# Step 2: Check missing values
print("\nMISSING VALUES:")
print(df.isnull().sum())

# ============================================
# PART 1: HANDLING MISSING VALUES
# ============================================

# Step 3: Mean imputation
df['Age'] = df['Age'].fillna(df['Age'].mean())

# Step 4: Median imputation
df['Salary'] = df['Salary'].fillna(df['Salary'].median())

# Step 5: Forward fill
df['Experience'] = df['Experience'].ffill()

print("\nDATA AFTER IMPUTATION:")
print(df)

# ============================================
# PART 2: CATEGORICAL ENCODING
# ============================================

# Step 6: Label encoding for Purchased
label_encoder = LabelEncoder()

df['Purchased'] = label_encoder.fit_transform(
    df['Purchased']
)

# Step 7: Ordinal encoding for Education
education_order = [
    'School',
    'Bachelor',
    'Master'
]

ordinal_encoder = OrdinalEncoder(
    categories=[education_order]
)

df['Education'] = ordinal_encoder.fit_transform(
    df[['Education']]
).astype(int)

# Step 8: One-hot encoding for City
df = pd.get_dummies(
    df,
    columns=['City'],
    dtype=int
)

# ============================================
# PART 3: DISPLAY PROCESSED DATA
# ============================================

print("\nFINAL PROCESSED DATASET:")
print(df)

# Step 9: Verify missing values
print("\nMISSING VALUES AFTER PROCESSING:")
print(df.isnull().sum())

# Step 10: Display dataset information
print("\nDATASET INFORMATION:")
df.info()
```
